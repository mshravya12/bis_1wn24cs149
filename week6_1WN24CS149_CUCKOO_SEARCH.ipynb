```python

import random
import math

# Cities and their coordinates
cities = [(0, 0), (2, 3), (5, 2), (6, 6), (1, 5)]

# Parameters
n = 10
Pa = 0.25
MaxT = 100

# Calculate distance between two cities
def distance(a, b):
    x = cities[a][0] - cities[b][0]
    y = cities[a][1] - cities[b][1]
    return math.sqrt(x*x + y*y)

# Calculate total route distance
def fitness(route):
    total = 0

    for i in range(len(route)):
        total += distance(route[i], route[(i+1) % len(route)])

    return total

# Generate random nests
nests = []

for i in range(n):
    route = list(range(len(cities)))
    random.shuffle(route)
    nests.append(route)

# Cuckoo Search Algorithm
for t in range(MaxT):

    # Generate new cuckoo using Levy Flight
    new_route = random.choice(nests).copy()

    steps = min(len(cities), int(random.paretovariate(1.5)))

    for i in range(steps):
        a, b = random.sample(range(len(cities)), 2)
        new_route[a], new_route[b] = new_route[b], new_route[a]

    # Evaluate new solution
    new_fitness = fitness(new_route)

    # Choose a random nest
    j = random.randint(0, n-1)

    # Compare fitness
    if new_fitness < fitness(nests[j]):
        nests[j] = new_route

    # Abandon worst nests
    nests.sort(key=fitness)

    count = int(Pa * n)

    for i in range(count):
        route = list(range(len(cities)))
        random.shuffle(route)
        nests[-(i+1)] = route

    # Keep the best solution
    nests.sort(key=fitness)

# Display best route
best = nests[0]

print("Best Route:")

for city in best:
    print(city + 1, end=" -> ")

print(best[0] + 1)

print("Shortest Distance:", round(fitness(best), 2))
```
