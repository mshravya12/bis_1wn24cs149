```python

import random

# Processing times of jobs
jobs = [5, 3, 8, 2, 4]

# Number of wolves and iterations
N = 10
iterations = 20

# Fitness function
def fitness(order):
    total = 0
    completion = 0

    for i in order:
        completion += jobs[i]
        total += completion

    return total

# Initialize wolves with random job orders
wolves = []

for i in range(N):
    order = list(range(len(jobs)))
    random.shuffle(order)
    wolves.append(order)

# GWO
for t in range(iterations):

    wolves.sort(key=fitness)

    alpha = wolves[0]
    beta = wolves[1]
    delta = wolves[2]

    new_wolves = [alpha[:], beta[:], delta[:]]

    for i in range(3, N):
        leaders = [alpha, beta, delta]
        new_order = []

        for j in range(len(jobs)):
            leader = random.choice(leaders)
            job = leader[j]

            if job not in new_order:
                new_order.append(job)

        # Add missing jobs
        for job in range(len(jobs)):
            if job not in new_order:
                new_order.append(job)

        new_wolves.append(new_order)

    wolves = new_wolves

# Display best solution
best = min(wolves, key=fitness)

print("Best Job Sequence:")

for i in best:
    print("J", i + 1, end=" ")

print("\nMinimum Total Completion Time:", fitness(best))
```
