```python

import random

# List of jobs (execution times)
jobs = [4, 6, 3, 5, 2, 7]

# Parameters
N = 10
MAX = 50
MACHINES = 2

# Calculate fitness (total completion time)
def fitness(solution):
    load = [0] * MACHINES

    for i in range(len(jobs)):
        load[solution[i]] += jobs[i]

    return max(load)


# Initialize cells randomly
cells = []

for i in range(N):
    solution = [random.randint(0, MACHINES - 1)
                for j in range(len(jobs))]
    cells.append(solution)


# Find the best cell
bestCell = cells[0]
bestFitness = fitness(bestCell)

for cell in cells:
    if fitness(cell) < bestFitness:
        bestFitness = fitness(cell)
        bestCell = cell[:]


# Main algorithm
for iteration in range(MAX):

    newCells = []

    # Process all cells
    for i in range(N):

        # Select a neighboring cell
        neighbor = cells[(i + 1) % N]

        # Copy current solution
        newSolution = cells[i][:]

        # Update solution using neighbor
        position = random.randint(0, len(jobs) - 1)
        newSolution[position] = neighbor[position]

        # Evaluate new solution
        if fitness(newSolution) < fitness(cells[i]):
            newCells.append(newSolution)
        else:
            newCells.append(cells[i][:])

    # Replace old cells
    cells = newCells

    # Find best solution
    for cell in cells:
        if fitness(cell) < bestFitness:
            bestFitness = fitness(cell)
            bestCell = cell[:]


# Display result
print("Job Execution Times:", jobs)
print("Best Job Assignment:", bestCell)
print("Minimum Completion Time:", bestFitness)

for m in range(MACHINES):
    assigned = [jobs[i] for i in range(len(jobs))
                if bestCell[i] == m]

    print("Machine", m + 1, "Jobs:", assigned)
    print("Machine", m + 1, "Total Time:", sum(assigned))
```
