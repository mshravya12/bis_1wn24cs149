```python
import numpy as np

def cost_function(x):
    cost = (x - 6)**2 + 10
    return cost

num_particles = 5
max_iterations = 6

w = 0.7
c1 = 1.5
c2 = 1.5

positions = np.random.uniform(0, 10, num_particles)

velocities = np.random.uniform(-1, 1, num_particles)

pbest_positions = positions.copy()

pbest_cost = cost_function(pbest_positions)

gbest_index = np.argmin(pbest_cost)

gbest_position = pbest_positions[gbest_index]

gbest_cost = pbest_cost[gbest_index]

for iteration in range(max_iterations):

    print("\n")
    print("========================================")
    print("ITERATION:", iteration + 1)
    print("========================================")

    current_cost = cost_function(positions)

    print("\nCurrent Positions:")
    print(positions)

    print("\nCurrent Costs:")
    print(current_cost)

    for i in range(num_particles):

        if current_cost[i] < pbest_cost[i]:

            pbest_positions[i] = positions[i]

            pbest_cost[i] = current_cost[i]

    print("\npBEST Positions:")
    print(pbest_positions)

    print("\npBEST Costs:")
    print(pbest_cost)

    best_index = np.argmin(pbest_cost)

    if pbest_cost[best_index] < gbest_cost:

        gbest_position = pbest_positions[best_index]

        gbest_cost = pbest_cost[best_index]

    print("\ngBEST Position:")
    print(gbest_position)

    print("gBEST Cost:")
    print(gbest_cost)

    r1 = np.random.random(num_particles)

    r2 = np.random.random(num_particles)

    velocities = (
        w * velocities
        + c1 * r1 * (pbest_positions - positions)
        + c2 * r2 * (gbest_position - positions)
    )

    print("\nNew Velocities:")
    print(velocities)

    positions = positions + velocities

    positions = np.clip(positions, 0, 10)

    print("\nNew Positions:")
    print(positions)

print("\n")
print("========================================")
print("FINAL RESULT")
print("========================================")

print("Best Delivery Center Location:",
      gbest_position, "km")

print("Minimum Delivery Cost:",
      gbest_cost)
```
